```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/weatherAUS.csv")

print("Original dataset shape:", df.shape)
df.head()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day

df.head()
```

```python
df = df.drop("Date", axis=1)
```

```python
columns_to_drop = ["Evaporation", "Sunshine", "Cloud9am", "Cloud3pm"]

df = df.drop(columns=[col for col in columns_to_drop if col in df.columns])

print("Shape after dropping high missing value columns:", df.shape)
```

```python
df = df.dropna(subset=["RainTomorrow"])

print("Shape after removing missing target rows:", df.shape)
```

```python
df["RainTomorrow"] = df["RainTomorrow"].map({"No": 0, "Yes": 1})

df["RainTomorrow"].value_counts()
```

```python
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_columns = df.select_dtypes(include=["object"]).columns.tolist()

print("Numeric Columns:")
print(numeric_columns)

print("\nCategorical Columns:")
print(categorical_columns)
```

```python
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

print("Numeric missing values handled.")
```

```python
for col in categorical_columns:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Categorical missing values handled.")
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
df.to_csv("../data/weatherAUS_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```

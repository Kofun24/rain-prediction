```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/weatherAUS.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
missing_values = df.isnull().sum().sort_values(ascending=False)
missing_values
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage.sort_values(ascending=False)
```

```python
df["RainTomorrow"].value_counts()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="RainTomorrow")
plt.title("Rain Tomorrow Count")
plt.xlabel("Rain Tomorrow")
plt.ylabel("Count")
plt.show()
```

```python
df["RainToday"].value_counts()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="RainToday")
plt.title("Rain Today Count")
plt.xlabel("Rain Today")
plt.ylabel("Count")
plt.show()
```

```python
df.describe()
```

```python
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns
numeric_columns
```

```python
categorical_columns = df.select_dtypes(include=["object"]).columns
categorical_columns
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["MinTemp"], kde=True)
plt.title("Minimum Temperature Distribution")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["MaxTemp"], kde=True)
plt.title("Maximum Temperature Distribution")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["Rainfall"], kde=True)
plt.title("Rainfall Distribution")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="RainTomorrow", y="Humidity3pm")
plt.title("Humidity at 3pm vs Rain Tomorrow")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="RainTomorrow", y="Pressure3pm")
plt.title("Pressure at 3pm vs Rain Tomorrow")
plt.show()
```

```python
plt.figure(figsize=(14, 10))
correlation = df.select_dtypes(include=["int64", "float64"]).corr()
sns.heatmap(correlation, cmap="coolwarm", annot=False)
plt.title("Correlation Heatmap")
plt.show()
```
